# Create Idealized Forcing 

```python
import numpy as np
import pickle as pkl
```

```python
folder = "./"
```

```python
# Idealized forcing, same definition as in notebooks/idealized/Idealized_*.ipynb.
# The idealized run starts from the end of the spinup (t = 0 at the spinup end).
t_end_ideal = 130*60*60*24
time_0 = 1698796800.0
t_bc = np.arange(0, t_end_ideal+3600,3600) # time points for boundary condition [s]
q_inf = 2e-8*np.ones_like(t_bc) # recharge [m/s]

t_bc_days = t_bc / 86400
T_cold, T_warm = -2., 2.0
t1, w1 = 30.0, 15.0   # cooling transition centered on day 30, ~30 d wide
t2, w2 = 90.0, 15.0   # warming transition centered on day 90, ~30 d wide
s_cool = 0.5 * (1 + np.tanh((t_bc_days - t1) / w1))
s_warm = 0.5 * (1 + np.tanh((t_bc_days - t2) / w2))
T_surface = 0.0 + (T_cold - 0.0) * s_cool + (T_warm - T_cold) * s_warm

# evaluate hourly with the same linear interpolation the simulation uses
T_ideal = T_surface
recharge_ideal = q_inf
time_array_ideal = time_0 + t_bc

db = {
    "dates": time_array_ideal,
    "time": t_bc,
    "T_surf": T_ideal,
    "recharge": recharge_ideal
}
with open(folder+"idealized_forcing.pkl", "wb") as f:
        pkl.dump(db, f)
```
